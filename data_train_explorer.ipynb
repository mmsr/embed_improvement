```python
import json
from decimal import Decimal
#extracted_file =  "train_extracted_improved.jsonl"
extracted_file =  "train_extracted.jsonl"

triplets = set()

with open(extracted_file, "r") as f:
    for line in f:
        r = json.loads(line, parse_float=Decimal)

        key = (
            r["number"],
            r["positive_number"],
            r["negative_number"]
        )
        triplets.add(key)   # set → removes duplicates automatically
```

```python
sorted_triplets = sorted(triplets)
```

```python
with open("dedup_sorted_original.jsonl", "w") as f:
    for n, p, neg in sorted_triplets:
        f.write(json.dumps({
            "number": str(n),
            "positive_number": str(p),
            "negative_number": str(neg)
        }) + "\n")
```

```python
import random

sample_size = 5000

# if triplets already created as shown before
triplets_list = list(triplets)

sampled = random.sample(triplets_list, min(sample_size, len(triplets_list)))
```
